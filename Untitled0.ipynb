```python
import numpy as np
import matplotlib.pyplot as plt
def eigen_decomposition(A):
  if A.shape[0] == A.shape[1]:
    w, V = np.linalg.eig(A)
    m = []
    for i in range(len(w)):
      v = V[:, i]
      lam = w[i]
      a = np.allclose(A @ v, lam*v)
      m.append(a)

    if all(m):
      print("все вірно")
    else:
      print("є помилки")
  else:
      raise ValueError("Error")

  b = np.allclose(w.imag, 0)

  if b:
    w = np.real(w)

  return w, V
```

```python
M1 = np.array([[2, 1],
               [1, 2]])
w, V = eigen_decomposition(M1)
print(w)
print(V)
```

```python
M2 = np.array([[0, -1],
               [1, 0]])
w, V = eigen_decomposition(M2)
print(w)
print(V)
```

```python
M3 = np.array([[1, 2, 3],
               [4, 5, 6]])
try:
    w, V = eigen_decomposition(M3)
    print(w)
    print(V)
except ValueError as e:
  print("Error", e)
```

Для симетричної матриці [[2, 1], [1, 2]] функція повернула значення 3 та 1. Обидва значення вийшли повністю дійсними, без уявної частини. Отримані два власні вектори виявилися взаємно перпендикулярними, їхній скалярний добуток дорівнює нулю. Це наочно підтверджує теоретичну властивість ортогональності власних векторів для будь-якої дійсної симетричної матриці.
Для матриці повороту власні значення вийшли суто комплексними: +i та -i. Геометрично це логічно, тому що поворот на площині розвертає кожен вектор, і жоден дійсний вектор для повороту на 90 градусів не може зберегти свій початковий напрямок. Функцію np.real тут застосовувати не можна. В умові лабораторної np.real потрібна тільки для того, щоб прибирати мікроскопічні похибки обчислень вигляду a + 0*i. У випадку матриці повороту уявна частина не є нульовою і несе ключовий математичний зміст, тому її відкидання обнулить значення та спотворить результат.
Для неквадратної матриці функція видала помилку розмірності. Власні значення та власні вектори існують тільки для квадратних операторів, оскільки неквадратна матриця відображає вектор у простір іншої розмірності, де перевіряти колінеарність до і після перетворення фізично неможливо.
Співвідношення A @ v = lambda * v ми перевірили саме чисельно через np.allclose. Важливо підкреслити, що це практична чисельна перевірка з урахуванням похибки машинних обчислень, а не строге аналітичне доведення.

```python
import os
import urllib.request
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
from sklearn.decomposition import PCA

if not os.path.exists("/content/58c09721eeaeda9c078c26706c8b0d7d.jpg"):
    url = "https://upload.wikimedia.org/wikipedia/commons/thumb/c/c4/Times_Square_New_York_City_HDR_2013.jpg/800px-Times_Square_New_York_City_HDR_2013.jpg"
    req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
    with urllib.request.urlopen(req) as response, open("/content/58c09721eeaeda9c078c26706c8b0d7d.jpg", 'wb') as out_file:
        out_file.write(response.read())


image_raw = imread("/content/58c09721eeaeda9c078c26706c8b0d7d.jpg")
print("Розмір початкового зображення (висота, ширина, канали):", image_raw.shape)

plt.figure(figsize=(8, 6))
plt.imshow(image_raw)
plt.title(f"Original Color Image {image_raw.shape}")
plt.show()


image_sum = image_raw.sum(axis=2)
print("Розмір Ч/Б зображення (висота, ширина):", image_sum.shape)

image_bw = image_sum / image_sum.max()
print("Максимальне значення яркості після нормалізації:", image_bw.max())

plt.figure(figsize=(8, 6))
plt.imshow(image_bw, cmap='gray')
plt.title(f"Black & White Image {image_bw.shape}")
plt.show()
```

```python
pca = PCA()
pca.fit(image_bw)

cumulative_variance = np.cumsum(pca.explained_variance_ratio_) * 100

k = np.argmax(cumulative_variance >= 95) + 1
print(f"Number of components explaining 95% variance: {k}")

plt.figure(figsize=(10, 5))
plt.plot(cumulative_variance)
plt.axhline(y=95, color='r', linestyle='--')
plt.axvline(x=k-1, color='k', linestyle='--')
plt.xlabel("Principal components")
plt.ylabel("Cumulative Explained variance")
plt.title("Cumulative Explained Variance explained by the components")
plt.grid(True, linestyle=':')
plt.show()

pca_95 = PCA(n_components=k)
image_transformed = pca_95.fit_transform(image_bw)
image_reconstructed_95 = pca_95.inverse_transform(image_transformed)

plt.figure(figsize=(8, 6))
plt.imshow(image_reconstructed_95, cmap='gray', vmin=0, vmax=1)
plt.title(f"Reconstructed image with {k} components (95% variance)")
plt.show()
```

```python
components_list = [5, 15, 25, k, 75, 100, 170]

fig, axes = plt.subplots(2, 4, figsize=(16, 10))
axes = axes.flatten()

for i, n_comp in enumerate(components_list):
    n_comp_safe = min(n_comp, min(image_bw.shape))

    pca_temp = PCA(n_components=n_comp_safe)
    compressed = pca_temp.fit_transform(image_bw)
    reconstructed = pca_temp.inverse_transform(compressed)

    axes[i].imshow(reconstructed, cmap='gray', vmin=0, vmax=1)
    axes[i].set_title(f"Components: {n_comp_safe}")

plt.tight_layout()
plt.show()
```

Математично перші 63 компоненти забирають левову частку дисперсії, оскільки описують масштабні перепади яскравості між світлим небом, сонцем і темними скелями. За такого вибору повністю зникає мильність і зернистий шум на дрібних елементах дерева, але при цьому ми все одно зберігаємо значне стиснення (використовуємо лише 100 компонент із 474).

Перевірка алгоритму для повідомлення "Hello, World!" підтвердила стовідсоткову точність відновлення: вираз перевірки message == decrypted_message у коді повертає True. Для повної відтворюваності результату генерація випадкової матриці ключа виконувалася із зафіксованим значенням seed (через np.random.seed(42)).   Під час дешифрування з'являються залишкові комплексні хвости порядку 1e-11j. Причина в тому, що випадкова несиметрична матриця ключа зазвичай має комплексно спряжені пари власних значень, тому функція np.linalg.eig проводить обчислення у комплексному просторі. Математично уявна частина вектора символів точно дорівнює нулю, але через чисельні похибки заокруглення чисел із рухомою комою в пам'яті залишаються мікроскопічні уявні залишки. Саме тому на етапі розшифрування у функції decrypt_message ми беремо дійсну частину через np.real та округлюємо значення до цілих чисел перед викликом chr().   Зворотне перетворення базується на діагоналізації матриці ключа A = P D P^(-1). Матриця P^(-1) вже обчислена при початковому розкладі, а D^(-1) тривіально отримується заміною кожного діагонального власного значення lambda_i на 1 / lambda_i, тому один і той самий спектральний розклад використовується і для шифрування, і для розшифрування. Для дешифрування достатньо помножити A^(-1) = P D^(-1) P^(-1) на вектор шифротексту.
